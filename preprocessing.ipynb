```
Preprocessing in Machine Learning is the process of cleaning and transforming raw data into a format that a machine learning model can use effectively.

Why is preprocessing important?

Raw data often contains:

Missing values
Duplicate records
Incorrect or inconsistent data
Categorical (text) values
Features with different scales
Noise and outliers

Preprocessing improves data quality, which can lead to better model performance.

```

````python
%pip install scikit-learn
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
````

finding our missing values


````python
df.isnull().sum()
````

fill missing values


````python
df["cast"] = df["cast"].fillna("Unknown")
df["date_added"] = df["date_added"].fillna("Unknown")
df["duration"] = df["duration"].fillna(df["duration"].mode()[0])
````

````python
df.isnull().sum()
````

Convert dates 

````python

df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
df["date_added"].dtype
````

Feature engineering

the process of transforming raw, messy data into meaningful, model-ready inputs

````python
df["added_year"] = df["date_added"].dt.year
df["added_month"] = df["date_added"].dt.month
df["genre_count"] = df["listed_in"].str.count(",") + 1
````

Encode target


````python
encoder=LabelEncoder()
df["type"] = encoder.fit_transform(df["type"])
````

Feature and target 

````python
X=df[["added_month", "added_year",  "genre_count"]]
````

Handling any mising numeric values

````python
X = X.fillna(X.mean())
````

Scale

````python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
````

split 

````python
y = df["type"]

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)
````
