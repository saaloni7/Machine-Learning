Regularization is a technique used in machine learning to prevent overfitting by adding a complexity penalty to the model's loss function, encouraging simpler models that perform well on new data. 

````python
import numpy as np 
w=np.array([2,5,-4])
lam=0.1
penalty=lam*np.sum(w**2)
print(penalty)
````

```
L1 and L2 regularization are methods used to prevent models from overfitting by adding a
 penalty to the loss function based on the size of the model's weights:
 L1 adds absolute values, L2 adds squared values, and both are controlled by a tuning hyperparameter. 
 ```

````python
import numpy as np 
w=np.array([2,5,-4])
lam=0.1
penalty=lam*np.sum(np.abs(w))
print(penalty)
````

$$
dw=
\frac{-2}{n}\sum x(y-\hat y)
+
2\lambda w
$$

````python
import numpy as np

X = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])

w = 0 #weight 
b = 0# bias

lr = 0.01# learnig rate 
epochs = 1000 #iteratins 

lam = 0.1

n = len(X)  

for epoch in range(epochs):

    pred = w*X+b

    dw = (-2/n)*np.sum(X*(y-pred))

    # L1 Regularization
    dw = dw + lam*np.sign(w)

    db = (-2/n)*np.sum(y-pred)

    w -= lr*dw
    b -= lr*db

print(w,b)
````
